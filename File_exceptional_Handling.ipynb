```python
#2. Write a Python program to read the contents of a file and print each line.
def read_and_print_file(filename):
    """
    Reads the contents of a file and prints each line.

    Args:
        filename (str): The name of the file to read.
    """
    try:
        with open(filename, 'r') as file:
            for line in file:
                print(line, end='')  # Print each line without adding an extra newline
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except IOError:
        print(f"Error: An error occurred while reading the file '{filename}'.")

if __name__ == "__main__":
    filename = input("Enter the name of the file to read: ")
    read_and_print_file(filename)
```

```python
#4. Write a Python script that reads from one file and writes its content to another file.
def copy_file_content(source_file, destination_file):
    """
    Reads the content of a source file and writes it to a destination file.

    Args:
        source_file (str): The path to the file to read from.
        destination_file (str): The path to the file to write to.
    """
    try:
        with open(source_file, 'r') as infile, open(destination_file, 'w') as outfile:
            content = infile.read()
            outfile.write(content)
        print(f"Successfully copied content from '{source_file}' to '{destination_file}'")
    except FileNotFoundError:
        print(f"Error: Source file '{source_file}' not found.")
    except IOError as e:
        print(f"Error: An IOError occurred: {e}")
        # Consider logging the error for more detailed tracking
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        #It's good practice to catch general exceptions, but you might want to log them.

if __name__ == "__main__":
    source_file = input("Enter the path to the source file: ")
    destination_file = input("Enter the path to the destination file: ")
    copy_file_content(source_file, destination_file)
```

```python
#5. How would you catch and handle division by zero error in Python?
def divide(numerator, denominator):
    """
    Divides two numbers and handles the case where the denominator is zero.

    Args:
        numerator (float): The numerator.
        denominator (float): The denominator.

    Returns:
        float: The result of the division, or None if the denominator is zero.
    """
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero. Returning None.")
        return None  # Explicitly return None

def main():
    """
    Main function to demonstrate division with error handling.
    """
    num1 = float(input("Enter the numerator: "))
    num2 = float(input("Enter the denominator: "))
    result = divide(num1, num2)

    if result is not None:
        print(f"{num1} / {num2} = {result}")
    else:
        print("Division was not performed due to a zero denominator.")

if __name__ == "__main__":
    main()
```

```python
#7. How do you log information at different levels (INFO, ERROR, WARNING) in Python using the logging module.
import logging

def configure_logging(log_file="application.log", level=logging.INFO):
    """
    Configures the logging system.

    Args:
        log_file (str): The name of the log file.
        level (int): The logging level (e.g., logging.INFO, logging.ERROR).
    """
    logging.basicConfig(
        filename=log_file,
        level=level,
        format='%(asctime)s - %(levelname)s - %(message)s'
    )

def do_something(x, y):
    """
    Simulates a function that performs an operation and logs at different levels.

    Args:
        x (int): The first number.
        y (int): The second number.
    """
    logging.info(f"do_something called with x={x}, y={y}")  # Log INFO

    if y == 0:
        logging.error("Attempted division by zero")  # Log ERROR
        return None  # Important:  Return None or raise an exception

    if x < 0 or y < 0:
        logging.warning(f"Negative input values: x={x}, y={y}") # Log WARNING.
        result = x / y
        return result

    result = x / y
    logging.info(f"Result of division: {result}")
    return result

def main():
    """
    Main function to demonstrate logging at different levels.
    """
    configure_logging()  # Configure logging to application.log, INFO level

    print("Demonstrating logging levels...")
    do_something(10, 2)
    do_something(10, 0)
    do_something(-5, 5)
    do_something(20,4)

    logging.critical("Program execution finished.") # Log CRITICAL
    print("Check 'application.log' for log messages.")

if __name__ == "__main__":
    main()
```

```python
#8. Write a program to handle a file opening error using exception handling.
def open_file_safely(filename, mode='r'):
    """
    Opens a file and handles potential errors.

    Args:
        filename (str): The name of the file to open.
        mode (str, optional): The mode in which to open the file ('r', 'w', 'a', etc.).
            Defaults to 'r' (read mode).

    Returns:
        file object: The opened file object if successful, None otherwise.
    """
    try:
        file = open(filename, mode)
        return file  # Return the file object on success
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None  # Explicitly return None to indicate failure
    except PermissionError:
        print(f"Error: Permission denied to open file '{filename}' in mode '{mode}'.")
        return None
    except OSError as e:
        print(f"Error: An error occurred while opening file '{filename}': {e}")
        return None
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        return None

def read_from_file(filename):
    """Reads from file safely"""
    file = open_file_safely(filename, 'r')
    if file:
        with file: # Use file object returned
            content = file.read()
            print(content)

def write_to_file(filename, text):
    """Writes to a file safely"""
    file = open_file_safely(filename, 'w')
    if file:
        with file: # Use file object returned
            file.write(text)

def main():
    """
    Main function to demonstrate file opening error handling.
    """
    read_filename = input("Enter the name of the file to read: ")
    read_from_file(read_filename)

    write_filename = input("Enter the name of the file to write to: ")
    text_to_write = "This is a test string.\nAdding more content."
    write_to_file(write_filename, text_to_write)



if __name__ == "__main__":
    main()
```

```python
#9. How can you read a file line by line and store its content in a list in Python?
def read_file_to_list(filename):
    """
    Reads a file line by line and stores the content in a list.

    Args:
        filename (str): The name of the file to read.

    Returns:
        list: A list where each element is a line from the file, or None on error.
    """
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()  # Read all lines into a list
        return lines
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    except IOError as e:
        print(f"Error reading file '{filename}': {e}")
        return None

def main():
    """
    Main function to demonstrate reading a file into a list.
    """
    filename = input("Enter the name of the file to read: ")
    lines = read_file_to_list(filename)

    if lines:
        print(f"Content of '{filename}' as a list:")
        for i, line in enumerate(lines):
            print(f"Line {i+1}: {line.strip()}")  # Use strip() to remove extra newlines
    else:
        print(f"Could not read file")

if __name__ == "__main__":
    main()
```

```python
#10. How can you append data to an existing file in Python?
def append_to_file(filename, data):
    """
    Appends data to an existing file.

    Args:
        filename (str): The name of the file to append to.
        data (str): The data to append.
    """
    try:
        with open(filename, 'a') as file:
            file.write(data)
        print(f"Successfully appended data to '{filename}'")
    except IOError as e:
        print(f"Error: An error occurred while appending to file '{filename}': {e}")

def main():
    """
    Main function to demonstrate appending data to a file.
    """
    filename = input("Enter the name of the file to append to: ")
    data_to_append = input("Enter the data to append: ")
    append_to_file(filename, data_to_append)

if __name__ == "__main__":
    main()
```

```python
#12. Write a program that demonstrates using multiple except blocks to handle different types of exceptions.
def process_data(data):
    """
    Processes the given data, handling potential errors.

    Args:
        data: The data to process (can be a list, tuple, or dictionary).
    """
    try:
        if not data:
            raise ValueError("Data cannot be empty.")

        if isinstance(data, list):
            element = data[2]  # Accesses the third element (index 2)
            result = 10 / element
        elif isinstance(data, tuple):
            element = data[5] #Accesses the sixth element
            result = "abc" + element
        elif isinstance(data, dict):
            element = data["key"]  # Accesses the value associated with the key "key"
            result = element * 2
        else:
            raise TypeError("Unsupported data type.  Must be list, tuple, or dict.")
        print(f"Processed data: {result}")

    except IndexError:
        print("Error: Index out of range.  The data structure is too short.")
    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")
    except TypeError as e:
        print(f"Error: Type error: {e}")
    except KeyError:
        print("Error: Key 'key' not found in the dictionary.")
    except ValueError as e:
        print(f"Error: Value error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}") # Added a catch-all.
    else:
        print("Data processed successfully.") # Added an else block
    finally:
        print("Processing complete.") # Added a finally block

def main():
    """
    Main function to demonstrate handling multiple exception types.
    """
    print("Demonstrating multiple except blocks...")

    list_data = [1, 2, 3, 4, 5]
    tuple_data = (10, 20, 30, 40, 50, "hello")
    dict_data = {"name": "Alice", "age": 30, "city": "New York", "key": 123}
    empty_list = []
    string_data = "abc"

    process_data(list_data)
    process_data(tuple_data)
    process_data(dict_data)
    process_data(empty_list)
    process_data(string_data)

    print("End of program.") #Added to show the end.

if __name__ == "__main__":
    main()
```

```python
#14. Write a program that uses the logging module to log both informational and error messages.
import logging

def configure_logging(log_file="application.log"):
    """
    Configures the logging system to log to a file.

    Args:
        log_file (str, optional): The name of the log file. Defaults to "application.log".
    """
    logging.basicConfig(
        filename=log_file,
        level=logging.INFO,  # Set the logging level to INFO (or higher)
        format='%(asctime)s - %(levelname)s - %(message)s'
    )

def process_data(data):
    """
    Processes the given data and logs informational and error messages.

    Args:
        data: The data to process.
    """
    logging.info("Starting to process data...")  # Log informational message

    try:
        if not data:
            raise ValueError("Input data is empty.")  # Raise an error for empty data

        if not isinstance(data, list):
            raise TypeError("Input data must be a list.")

        for item in data:
            logging.info(f"Processing item: {item}")  # Log info about each item
            if not isinstance(item, (int, float)):
                raise ValueError(f"Invalid data type: {type(item)}.  Expected int or float, got {type(item)}.")
            result = item * 2
            logging.info(f"Result of processing {item}: {result}")

    except ValueError as e:
        logging.error(f"ValueError: {e}")  # Log the error with detailed information
        return None  # Return None to indicate failure
    except TypeError as e:
        logging.error(f"TypeError: {e}")
        return None
    except Exception as e:
        logging.error(f"An unexpected error occurred: {e}")
        return None
    else:
        logging.info("Data processing completed successfully.")
        return [item * 2 for item in data] # return the processed data
    finally:
        logging.info("Finished processing data.")  # Log message

def main():
    """
    Main function to demonstrate logging informational and error messages.
    """
    configure_logging()  # Configure the logging system

    print("Demonstrating logging...")

    # Example 1: Successful data processing
    data1 = [1, 2, 3, 4, 5]
    processed_data1 = process_data(data1)
    if processed_data1:
        print(f"Processed data 1: {processed_data1}")

    # Example 2: Error due to empty data
    data2 = []
    processed_data2 = process_data(data2)
    if processed_data2 is None:
        print("Data processing failed for data 2.")

    # Example 3: Error due to invalid data type in list
    data3 = [1, 2, "a", 4, 5]
    processed_data3 = process_data(data3)
    if processed_data3 is None:
        print("Data processing failed for data 3.")

    data4 = [10,20,30,40]
    processed_data4 = process_data(data4)
    if processed_data4:
        print(f"Processed data 4: {processed_data4}")

    print("Check 'application.log' for log messages.")

if __name__ == "__main__":
    main()
```

```python
#17.  Write a Python program to create and write a list of numbers to a file, one number per line.
def write_numbers_to_file(numbers, filename="numbers.txt"):
    """
    Writes a list of numbers to a file, one number per line.

    Args:
        numbers (list): A list of numbers (integers or floats).
        filename (str, optional): The name of the file to write to.
            Defaults to "numbers.txt".
    """
    try:
        with open(filename, 'w') as file:
            for number in numbers:
                file.write(str(number) + "\n")
        print(f"Successfully wrote numbers to {filename}")
    except Exception as e:
        print(f"An error occurred: {e}")

def create_number_list(n, start=0, step=1):
    """
    Creates a list of numbers.

    Args:
        n (int): The number of elements in the list.
        start (int or float, optional): The starting value. Defaults to 0.
        step (int or float, optional): The step between consecutive numbers. Defaults to 1.

    Returns:
        list: A list of numbers.  Returns an empty list if n is not positive.
    """
    if n <= 0:
        return []
    return [start + i * step for i in range(n)]

def main():
    """
    Main function to demonstrate the usage of the functions.
    """
    # Create a list of numbers
    num_elements = 10
    start_value = 1
    step_value = 2
    number_list = create_number_list(num_elements, start_value, step_value)
    print(f"Generated list: {number_list}")

    # Write the list to a file
    file_name = "my_numbers.txt"
    write_numbers_to_file(number_list, file_name)

    # Create another list with floats and write it to a file
    float_list = [3.14, 2.71, 1.618, 0.577, 1.414]
    write_numbers_to_file(float_list, "float_numbers.txt")

if __name__ == "__main__":
    main()
```

```python
#21. Write a Python program that reads a file and prints the number of occurrences of a specific word?
import re  # Import the regular expression module

def count_word_occurrences(filename, word):
    """
    Reads a file and counts the number of occurrences of a specific word.
    The count is case-insensitive, and punctuation is ignored.

    Args:
        filename (str): The name of the file to read.
        word (str): The word to count.

    Returns:
        int: The number of times the word appears in the file.
             Returns 0 if the file does not exist or an error occurs.
    """
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            text = file.read()
    except FileNotFoundError:
        print(f"Error: File not found - {filename}")
        return 0
    except Exception as e:
        print(f"An error occurred while reading {filename}: {e}")
        return 0

    # Use a regular expression to find all occurrences of the word, case-insensitively
    # \b ensures that we are matching whole words (e.g., "the" not "there").
    pattern = r"\b" + re.escape(word) + r"\b"  # Escape the word for special characters
    matches = re.findall(pattern, text, re.IGNORECASE)
    return len(matches)

def main():
    """
    Main function to demonstrate the usage of the count_word_occurrences function.
    """
    # Create a sample text file
    with open("sample.txt", "w", encoding='utf-8') as f:
        f.write("The quick brown fox jumps over the lazy dog.  The dog is lazy.  THE END.\n")
        f.write("the quick brown fox.\n")

    # Count occurrences of "the"
    word_to_count = "the"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Count occurrences of "dog"
    word_to_count = "dog"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Count occurrences of "nonexistent"
    word_to_count = "nonexistent"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Example with a file that doesn't exist
    count = count_word_occurrences("missing_file.txt", "word")
    print(f"Count in missing_file: {count}") # Should print 0

if __name__ == "__main__":
    main()
```

```python
#22. How can you check if a file is empty before attempting to read its contents?
import re  # Import the regular expression module
import os

def count_word_occurrences(filename, word):
    """
    Reads a file and counts the number of occurrences of a specific word.
    The count is case-insensitive, and punctuation is ignored.

    Args:
        filename (str): The name of the file to read.
        word (str): The word to count.

    Returns:
        int: The number of times the word appears in the file.
             Returns 0 if the file does not exist or an error occurs.
    """
    try:
        if os.path.getsize(filename) == 0:
            print(f"Error: File is empty - {filename}")
            return 0

        with open(filename, 'r', encoding='utf-8') as file:
            text = file.read()
    except FileNotFoundError:
        print(f"Error: File not found - {filename}")
        return 0
    except Exception as e:
        print(f"An error occurred while reading {filename}: {e}")
        return 0

    # Use a regular expression to find all occurrences of the word, case-insensitively
    # \b ensures that we are matching whole words (e.g., "the" not "there").
    pattern = r"\b" + re.escape(word) + r"\b"  # Escape the word for special characters
    matches = re.findall(pattern, text, re.IGNORECASE)
    return len(matches)

def main():
    """
    Main function to demonstrate the usage of the count_word_occurrences function.
    """
    # Create a sample text file
    with open("sample.txt", "w", encoding='utf-8') as f:
        f.write("The quick brown fox jumps over the lazy dog.  The dog is lazy.  THE END.\n")
        f.write("the quick brown fox.\n")

    # Create an empty file
    open("empty.txt", "w").close()

    # Count occurrences of "the"
    word_to_count = "the"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Count occurrences of "dog"
    word_to_count = "dog"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Count occurrences of "nonexistent"
    word_to_count = "nonexistent"
    count = count_word_occurrences("sample.txt", word_to_count)
    print(f"The word '{word_to_count}' appears {count} times in the file.")

    # Example with a file that doesn't exist
    count = count_word_occurrences("missing_file.txt", "word")
    print(f"Count in missing_file: {count}") # Should print 0

    # Example with an empty file
    count = count_word_occurrences("empty.txt", "the")
    print(f"Count in empty file: {count}")  # Should print 0

if __name__ == "__main__":
    main()
```

```python
#23. Write a Python program that writes to a log file when an error occurs during file handling.
import logging
import os

def read_file_and_count_words(filename, word_to_count):
    """
    Reads a file and counts the occurrences of a specific word, handling errors and logging them.

    Args:
        filename (str): The name of the file to read.
        word_to_count (str): The word to count.

    Returns:
        int: The number of times the word appears in the file.
             Returns 0 if the file does not exist or an error occurs.
    """
    # Configure logging
    logging.basicConfig(filename='file_errors.log', level=logging.ERROR,
                        format='%(asctime)s - %(levelname)s - %(message)s')

    try:
        if not os.path.exists(filename):
            logging.error(f"File not found: {filename}")
            return 0

        if os.path.getsize(filename) == 0:
            logging.error(f"File is empty: {filename}")
            return 0

        with open(filename, 'r', encoding='utf-8') as file:
            text = file.read()  # Read the entire file content

        # Perform word counting (case-insensitive)
        count = text.lower().split().count(word_to_count.lower())
        return count

    except Exception as e:
        logging.error(f"An error occurred while processing the file: {e}")
        return 0

def main():
    """
    Main function to demonstrate file reading and error logging.
    """
    # Create a sample file
    with open("my_file.txt", "w", encoding="utf-8") as f:
        f.write("This is a sample file.  The word 'sample' appears once.\n")
        f.write("Sample text.\n")

    # Test case 1: Successful file read and word count
    word_count = read_file_and_count_words("my_file.txt", "sample")
    print(f"The word 'sample' appears {word_count} times in my_file.txt")

    # Test case 2: File not found
    word_count = read_file_and_count_words("nonexistent_file.txt", "word")
    print(f"The word 'word' appears {word_count} times in nonexistent_file.txt")  # This will print 0 and log an error

    # Test case 3: Empty file
    open("empty_file.txt", "w").close()  # Create an empty file
    word_count = read_file_and_count_words("empty_file.txt", "word")
    print(f"The word 'word' appears {word_count} times in empty_file.txt")  # This will print 0 and log an error

    print("Check 'file_errors.log' to see the logged errors.")

if __name__ == "__main__":
    main()
```
