1. What is the difference between a function and a method in Python?

 A function is a standalone block of code defined independently, while method is a function that is associated with an object.

```python
#Function
def add(x,y):
  return x + y

result = add(3,5)
print(result)
```

```python
#method
class Calculator:
  def add(self,x,y):
    return x + y

calc = Calculator()
result = calc.add(3,5)
print(result)
```

6. Explain the concept of generators in Python and how they are defined.

Generator is a special type of function that produces a sequence of a values using the yield keyword instead of return keyword, allowing you to iterate over them one at at time, rather than storing them all in memory at once.

```python
#Generator
def square_n(n):
  for i in range(n):
    yield i**2
gen = square_n(20)
gen
next(gen)
next(gen)
next(gen)
```

9. Explain the purpose and usage of the `map()` function in Python.

Map() function serves to apply a given function to each item of an iterable (such as a list or tuple) and returns an iterator that yields the results. It provides a concise way to perform operations on sequences without explicit loops.

```python

def square(n):
  return n*n
numbers = [1,2,3]
squared = map(square,numbers)
list(squared)
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

 All three are built-in functions in Python that operate on iterables like list,tuples etc and apply a function to them.
- map() applies a given function to each item in iterable and returns a new iterator with the results. It transforms each element indvidually.
- reduce() applies a function cumulatively to items of an iterable, from left to right, to reduce the iterable to a single value. it aggregates elements.
 - filter() constructs an iterator from elements of an iterable for which a function returns true. It selects elements based on a condition.

```python

from functools import reduce

numbers = [1,2,3,4,5]

#map()
squared = map(lambda x: x**2,numbers)
print(list(squared))

#filter()
even = filter(lambda x: x%2 == 0,numbers)
print(list(even))

#reduce()
sum = reduce(lambda x,y: x+y,numbers)
print(sum)
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(number):
    if number < 2:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True

for number in range(1, 201):
    if is_prime(number):
        print(number)
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, n_terms):
        self.n_terms = n_terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n_terms:
            raise StopIteration
        fib_number = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return fib_number

fib_iter =  FibonacciIterator(10)
for num in fib_iter:
  print(num,end= " ")
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

```python
def rn(n):
  for i in range(1,n+1):
    a = 2**i
    yield a

for val in rn(6):
    print(val)
```

7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            yield line.strip()

# Example usage
file_path = "Sample.txt"
for line in read_file_line_by_line(file_path):
    print(line)
```
